```python
import torch
from unsloth import FastLanguageModel
from transformers import TextStreamer, AutoTokenizer
import sys
import types
from transformers import modeling_utils

# Monkey patch for older transformers (if needed)
def shard_checkpoint(state_dict, max_shard_size, weights_name):
    return [(state_dict, weights_name)]

if not hasattr(modeling_utils, 'shard_checkpoint'):
    modeling_utils.shard_checkpoint = shard_checkpoint
    print("✓ Monkey patched shard_checkpoint")

# ------------------------------------------------------------
# 1. Load the full model from the saved directory
# ------------------------------------------------------------
saved_model_path = "outputs/agent-model"  # Your saved model path

print(f"Loading model from {saved_model_path}...")
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name=saved_model_path,
    max_seq_length=2048,
    load_in_4bit=True,          # Keep 4-bit quantization
    device_map="cuda",          # Force all layers to GPU
    ignore_mismatched_sizes=True,  # Allow embedding size mismatch (saved vs config)
)

# ------------------------------------------------------------
# 2. Prepare model for inference
# ------------------------------------------------------------
model.eval()
FastLanguageModel.for_inference(model)  # Enable fast inference

# ------------------------------------------------------------
# 3. Generation function
# ------------------------------------------------------------
def generate_response(prompt: str, max_new_tokens=128, temperature=0.7):
    inputs = tokenizer(prompt, return_tensors="pt", truncation=True, max_length=2048)
    inputs = {k: v.to(model.device) for k, v in inputs.items()}
    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            temperature=temperature,
            do_sample=True,
            top_p=0.9,
            pad_token_id=tokenizer.pad_token_id,
            eos_token_id=tokenizer.eos_token_id,
        )
    input_len = inputs["input_ids"].shape[1]
    response_ids = outputs[0][input_len:]
    response = tokenizer.decode(response_ids, skip_special_tokens=True)
    return response.strip()

# ------------------------------------------------------------
# 4. Interactive loop
# ------------------------------------------------------------
print("Model ready! Enter your prompts (type 'quit' to exit).")
while True:
    user_input = input("\nUser: ")
    if user_input.lower() in ["quit", "exit"]:
        break
    response = generate_response(user_input)
    print(f"Assistant: {response}")
```
