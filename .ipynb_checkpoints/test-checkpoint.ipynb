```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)
from transformers.utils import logging as hf_logging

hf_logging.set_verbosity_error()


import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
from peft import PeftModel
import transformers.modeling_utils
if not hasattr(transformers.modeling_utils, 'shard_checkpoint'):
    def dummy_shard_checkpoint(*args, **kwargs):
        return {}, {}
    transformers.modeling_utils.shard_checkpoint = dummy_shard_checkpoint
    print("✅ Applied patch 1: added shard_checkpoint")
# =========================
# CONFIG
# =========================
BASE_MODEL = "microsoft/Phi-3.5-mini-instruct"
ADAPTER_PATH = "./results/checkpoint-1000"  # change if needed

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"


# =========================
# LOAD MODEL (SIMPLE + STABLE)
# =========================
def load_model():
    print("🔄 Loading tokenizer...")

    # Always load tokenizer from adapter if available
    # tokenizer = AutoTokenizer.from_pretrained(
    #     ADAPTER_PATH,
    #     trust_remote_code=True
    # )

    tokenizer = AutoTokenizer.from_pretrained(
    BASE_MODEL,
    trust_remote_code=True
)

    # 🔥 manually add your tokens again
    special_tokens = {
        "additional_special_tokens": [
            "<tool_use>", "</tool_use>",
            "<tool_name>", "</tool_name>",
            "<parameters>", "</parameters>",
        ]
    }
    tokenizer.add_special_tokens(special_tokens)

    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token

    print("🔄 Loading base model (no quantization for stability)...")

    # 🔥 REMOVE 4-bit → eliminates 80% issues
    base_model = AutoModelForCausalLM.from_pretrained(
        BASE_MODEL,
        torch_dtype=torch.float16 if DEVICE == "cuda" else torch.float32,
        trust_remote_code=True
    ).to(DEVICE)

    # 🔥 Match embeddings
    base_model.resize_token_embeddings(len(tokenizer))

    print(tokenizer.convert_tokens_to_ids("<tool_use>"))

    print("🔄 Loading LoRA adapter...")
    model = PeftModel.from_pretrained(base_model, ADAPTER_PATH)

    model.eval()

    print("✅ Model loaded successfully!\n")
    return model, tokenizer


# =========================
# PROMPT
# =========================
def build_prompt(user_input: str):
    return f"""### Instruction:
{user_input}

### Response:
"""


# =========================
# GENERATION (SAFE)
# =========================
def generate(model, tokenizer, text):
    prompt = build_prompt(text)

    inputs = tokenizer(prompt, return_tensors="pt").to(DEVICE)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=150,
            do_sample=True,
            temperature=0.7,
            top_p=0.9,
            use_cache=False,  # 🔥 CRITICAL FIX
            pad_token_id=tokenizer.eos_token_id,
        )

    decoded = tokenizer.decode(outputs[0], skip_special_tokens=False)

    if "### Response:" in decoded:
        decoded = decoded.split("### Response:")[1]

    return decoded.strip()


# =========================
# QUICK TESTS
# =========================
def run_tests():
    model, tokenizer = load_model()

    test_queries = [
        "Search latest AI trends",
        "Read configuration file and summarize it",
        "Compare cost of living in Bangalore and Tokyo",
    ]

    for q in test_queries:
        print(f"\n👤 {q}")
        try:
            response = generate(model, tokenizer, q)
            print(f"🤖 {response}")
        except Exception as e:
            print(f"❌ Error: {e}")


# =========================
# INTERACTIVE CHAT
# =========================
def chat():
    model, tokenizer = load_model()

    print("\n💬 Chat started (type 'exit' to quit)\n")

    while True:
        user_input = input("👤 You: ")

        if user_input.lower() in ["exit", "quit"]:
            break

        try:
            response = generate(model, tokenizer, user_input)
            print("\n🤖 Agent:\n", response)
            print("-" * 50)
        except Exception as e:
            print(f"❌ Error: {e}")


# =========================
# MAIN
# =========================
if __name__ == "__main__":
    run_tests()   # quick sanity check
    # chat()        # interactive
```
